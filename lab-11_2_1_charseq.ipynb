# lab 11_2_1 Character Sequence Processing

Character Sequence processing is an approach where text is processed at the individual character level. 

```python
import torch
import torch.optim as optim
import numpy as np
```

```python
# Random seed to make results deterministic and reproducible
torch.manual_seed(0)
```

```python
sample = " if you want you"
```

```python
# make dictionary
char_set = list(set(sample))
char_dic = {c: i for i, c in enumerate(char_set)}
print(char_dic)
```

```python
# hyper parameters
dic_size = len(char_dic)
hidden_size = len(char_dic)
learning_rate = 0.1
```

```python
# data setting

# Convert the sample text to indices using the character dictionary
sample_idx = [char_dic[c] for c in sample]

# input data
x_data = [sample_idx[:-1]]
x_one_hot = [np.eye(dic_size)[x] for x in x_data]

#target data
y_data = [sample_idx[1:]]

print(x_data)
print(x_one_hot)
print(y_data)
```

```python
# transform as torch tensor variable
X = torch.FloatTensor(x_one_hot)
Y = torch.LongTensor(y_data)
```

```python
# declare RNN
rnn = torch.nn.RNN(dic_size, hidden_size, batch_first=True)
```

```python
# loss & optimizer setting
criterion = torch.nn.CrossEntropyLoss()
optimizer = optim.Adam(rnn.parameters(), learning_rate)
```

```python
# start training
for i in range(51):
    optimizer.zero_grad()
    outputs, _status = rnn(X)
    loss = criterion(outputs.view(-1, dic_size), Y.view(-1))
    loss.backward()
    optimizer.step()

    result = outputs.data.numpy().argmax(axis=2)
    result_str = ''.join([char_set[c] for c in np.squeeze(result)])
    if i % 2 ==0:
        print(i, "loss: ", loss.item(), "prediction: ", result, "true Y: ", y_data, "prediction str: ", result_str)
```
